```python
import pandas as pd
import networkx as nx
from datetime import datetime, timedelta
import ast
```

```python
def load_data(file_path):
    columns_to_use = ['DOI', 'reference', 'earliest_pub_year']
    data = pd.read_csv(file_path, usecols=columns_to_use)
    return data
```

```python
def clean_data(data):  
    data = pd.read_csv("\\Users\\joshu\\Downloads\\UROP_Code\\Data\\climate_articles_unique_english.csv")

    if data['DOI'].isnull().any():
        print("There are missing values in the DOI column.")
    # Optionally, drop rows with missing DOI values
        data = data.dropna(subset=['DOI'])

# Step 2: Ensure all DOIs are strings and strip leading/trailing whitespaces
    data['DOI'] = data['DOI'].astype(str).str.strip()

# Step 3: Remove invalid DOIs (e.g., empty strings after cleaning)
    data = data[data['DOI'] != ""]

# Step 4: Check for duplicates and handle them
    if data['DOI'].duplicated().any():
        print("There are duplicate DOIs in the dataset.")
    # Optionally, drop duplicate DOIs (keep the first occurrence)
        data = data.drop_duplicates(subset=['DOI'])

# Step 5: Verify the cleaned DOI column
    print("Cleaned DOI column:")
    print(data['DOI'].head())
    return data
```

```python
def convert_year_to_datetime(year):
    try:
        year = int(year)  
        return datetime(year, 1, 1)  
    except (ValueError, TypeError):
        return None  


def process_references(ref):
    if pd.notna(ref):
        return [r.strip() for r in ref.split(';') if r.strip()]
    return []

def convert_reference_column(data):
    
    def safe_eval(ref):
        try:
            # Convert string representation of dictionary to an actual dictionary
            return ast.literal_eval(ref) if isinstance(ref, str) else ref
        except (ValueError, SyntaxError):
            # Log the issue and return None if the conversion fails
            print(f"Failed to parse reference: {ref}")
            return None

    # Apply the safe_eval function to the 'reference' column
    data['reference'] = data['reference'].apply(safe_eval)
    return data



def preprocess_data(data):
    
    data['time'] = data['earliest_pub_year'].apply(convert_year_to_datetime)

    data['reference'] = data['reference'].apply(process_references)

    data = data[pd.notna(data['time'])]

    return data
```

```python
def build_temporal_graph(data):

    # for _, row in data.iterrows():
    #     citing_doi = row['URL']
    #     references = row['reference']

    #     for ref in references:
    #         if isinstance(ref, str) and ref:  # Validate each reference
    #     # Match reference with DOI of other papers
    #             matching_row = data[data['URL'] == ref]
    #             if not matching_row.empty:
    #         # Add reference node with its time attribute
    #                 ref_time = matching_row.iloc[0]['time']
    #                 G.add_node(ref, time=ref_time)
    #                 G.add_edge(citing_doi, ref)

    G = nx.DiGraph()

    
    # Create a set of all DOIs in the dataset for quick lookup
    all_dois = set(data['DOI'])

    for _, row in data.iterrows():
        try:
            citing_doi = row['DOI']  # Current paper's DOI
            citing_time = row['time']  # Publication time of citing paper
            references = row['reference']  # Dictionary of references

            # Add the citing paper as a node with its 'time' attribute
            if pd.notna(citing_doi) and pd.notna(citing_time):
                G.add_node(citing_doi, time=citing_time)

            # Process references (assume the references column is a dictionary)
            if isinstance(references, dict):
                for ref_doi in references.values():  # Extract each DOI in the dictionary
                    if ref_doi in all_dois:  # Ensure the reference exists in the dataset
                        G.add_edge(citing_doi, ref_doi, time=citing_time)  # Edge with citation time
            else:
                print(f"References for DOI {citing_doi} are not in dictionary format: {references}")

        except Exception as e:
            print(f"Error processing row {row['DOI']}: {e}")
            continue

    print(f'Graph built: {len(G.nodes())} nodes, {len(G.edges())} edges')
    return G
```

```python
def remove_invalid_nodes(graph):
    invalid_nodes = [node for node, attrs in graph.nodes(data=True) if 'time' not in attrs or attrs['time'] is None]
    graph.remove_nodes_from(invalid_nodes)
    print(f"Removed {len(invalid_nodes)} invalid nodes.")
    return graph
```

```python
def calculate_cd_index(data, graph):
    cd_index_per_node = {}
    delta = timedelta(days=365 * 25)  

    for node in graph.nodes:
        try:
            node_time = graph.nodes[node].get('time')
            if node_time:
                cd_index_per_node[node] = float(nx.cd_index(graph, node, time_delta=delta, time="time"))
            else:
                cd_index_per_node[node] = None
        except Exception as e:
            print(f"Error calculating CD Index for node {node}: {e}")
            cd_index_per_node[node] = None

    return cd_index_per_node
```

```python
def main(file_path):
    data = load_data(file_path)
    data = clean_data(data)
    data = convert_reference_column(data)
    data = preprocess_data(data)

    graph = build_temporal_graph(data)
    graph = remove_invalid_nodes(graph)

    cd_indices = calculate_cd_index(data, graph)

    print(cd_indices)
    return cd_indices
```
